```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt
```

```python
#null: no linear relationship between house size and house price; b1=0
#alternative: there is significant relationship between house size and price; b1 !=0
```

```python
sacramento = pd.read_csv("../../01_materials/notebooks/dataset/sacramento.csv")
sacramento
```

```python
# Plot the relationship between x and y
plt.scatter(sacramento["sq__ft"], sacramento["price"], label='House size vs House price')
```

```python
#syntax: model (response variable ~ predictor, data), fit()
# Simple linear regression formula
model = smf.ols("price ~ sq__ft", data=sacramento).fit()
```

```python
print(model.summary())
```

```python
# Residual plots
fitted = model.fittedvalues
residuals = model.resid


# Homoscedasticity: Residuals vs. Fitted
plt.scatter(fitted, residuals)
plt.axhline(0, color='red', linestyle='--')
plt.xlabel('Fitted Values')
plt.ylabel('Residuals')
plt.title('Residuals vs Fitted')
plt.show()
```

```python
sm.qqplot(residuals, line = 'q')
plt.title('Q-Q Plot')
plt.show()
```

```python
#multivariable linear regression
multi_model = smf.ols("price ~ sq__ft + beds", data=sacramento).fit()
```

```python
print (multi_model.summary())
```

```python
sacramento_encoded = pd.get_dummies(sacramento, columns=["type"],drop_first=True)
sacramento_encoded
```

```python
#rename columns replace hyphens with underscores
sacramento_encoded.columns=sacramento_encoded.columns.str.replace("-","_")
sacramento_encoded
```

```python
# Now fit the model using the new column names
multi_cat_model = smf.ols("price ~ sq__ft + beds + type_Multi_Family + type_Residential", data=sacramento_encoded).fit()
```

```python
print(multi_cat_model.summary())
```
